```python

df['login_frequency'].fillna(df['login_frequency'].median(),inplace=True)
df['session_duration'].fillna(df['session_duration'].median(),inplace=True)
df['video_watch_time'].fillna(df['video_watch_time'].median(),inplace=True)
df['quiz_score'].fillna(df['quiz_score'].median(),inplace=True)
df['assignment_completion_rate'].fillna(df['assignment_completion_rate'].median(),inplace=True)
df['inactivity_days'].fillna(df['inactivity_days'].median(),inplace=True)
df['screen_time'].fillna(df['screen_time'].median(),inplace=True)
```

```python
df['device_type'].fillna(df['device_type'].mode()[0],inplace=True)
df['internet_quality'].fillna(df['internet_quality'].mode()[0],inplace=True)
df['study_environment'].fillna(df['study_environment'].mode()[0],inplace=True)
```

```python
df.isna().sum()
```

```python
df['focus_level'].value_counts()
```

```python
from sklearn.preprocessing import LabelEncoder
focus_encoder=LabelEncoder()
device_encoder=LabelEncoder()
internet_encoder=LabelEncoder()
study_encoder=LabelEncoder()
df['focus_level']=focus_encoder.fit_transform(df['focus_level'])
df['focus_level'].unique()
```

```python
list(focus_encoder.classes_)
```

```python
df['device_type']=device_encoder.fit_transform(df['device_type'])
df['internet_quality']=internet_encoder.fit_transform(df['internet_quality'])
df['study_environment']=study_encoder.fit_transform(df['study_environment'])
```

```python
x=df.drop('focus_level',axis=1)
y=df['focus_level']
```

```python
df.head()
```

```python
x.shape
```

```python
y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()

x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
knn=KNeighborsClassifier(n_neighbors=5)
knn.fit(x_train,y_train)
y_pred_knn=knn.predict(x_test)
score_knn=accuracy_score(y_test,y_pred_knn)
score_knn*100
```

```python
from sklearn.naive_bayes import GaussianNB
nb=GaussianNB()
nb.fit(x_train,y_train)
y_pred_nb=nb.predict(x_test)
score_nb=accuracy_score(y_test,y_pred_nb)
score_nb*100
```

```python
from sklearn.svm import SVC
svm=SVC()
svm.fit(x_train,y_train)
y_pred_svm=svm.predict(x_test)
score_svm=accuracy_score(y_test,y_pred_svm)
score_svm*100
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(criterion='entropy')
dt.fit(x_train,y_train)
y_pred_dt=dt.predict(x_test)
score_dt=accuracy_score(y_test,y_pred_dt)
score_dt*100
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(n_estimators=100)
rf.fit(x_train,y_train)
y_pred_rf=rf.predict(x_test)
score_rf=accuracy_score(y_test,y_pred_rf)
score_rf*100
```

```python
dff=pd.DataFrame({'Model':['KNN','NaiveBayes','SVM','Decision Tree','Random Forest'],
                  'Score':[score_knn*100,score_nb*100,score_svm*100,score_dt*100,score_rf*100]})
dff
```

```python
plt.bar(dff['Model'],dff['Score'])
plt.xlabel('Models')
plt.ylabel('Accuracy (%)')
plt.title('Model Comparison')
plt.show()
```

```python
print('Best Model : Random Forest')
print('Accuracy :',score_rf*100)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred_rf)
cm
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred_rf))
```
