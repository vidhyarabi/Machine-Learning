```python
x=df.iloc[:,:-1].values
y=df.iloc[:,-1].values
x
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
x_train
```

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaler.fit(x_train)
x_train=scaler.transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
y_pred
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
score=accuracy_score(y_test,y_pred)
score
```

```python
from sklearn.model_selection import GridSearchCV
model1=KNeighborsClassifier()
data={'n_neighbors':[3,5,7,9,11],'weights':['uniform','distance'],'metric':['euclidean','manhattan'],'algorithm':['ball_tree','kd_tree','brute']}
clf=GridSearchCV(model1,data,cv=10,scoring='accuracy')
clf.fit(x_train,y_train)
```

```python
print(clf.best_params_)
```

```python
model2=KNeighborsClassifier(n_neighbors=7,algorithm='ball_tree',metric= 'euclidean',weights='distance')
model2.fit(x_train,y_train)
y_pred1=model2.predict(x_test)
y_pred1
```

```python
score=accuracy_score(y_test,y_pred1)*100
score
```
