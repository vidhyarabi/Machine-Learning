```python
import nltk
nltk.download('stopwords')
nltk.download('punkt_tab')
from nltk.corpus import stopwords
word=stopwords.words('english')
from nltk.tokenize import word_tokenize
data1='''The Natural Language Toolkit (NLTK) is an open source Python library
    for Natural Language Processing.  A free online book is available'''
data2=word_tokenize(data1)
words1=[i for i in data2 if i.lower() not in word]
words1
```

```python
from nltk.stem import PorterStemmer
data=PorterStemmer()
words=['programming','programer','walking','reached','came','read','eated','swiming']
for i in words:
    print(i,':',data.stem(i))
```

```python
from nltk.stem import SnowballStemmer
data1=SnowballStemmer('english')
for i in words:
    print(i,':',data1.stem(i))
```

```python
# Lemmatization
nltk.download('omw-1.4')
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
data3=WordNetLemmatizer()
print('rocks',':',data3.lemmatize('rocks'))
print('reaches',':',data3.lemmatize('reaches'))
```
